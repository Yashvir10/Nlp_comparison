```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
nltk.download('punkt_tab')
```

```python
def preprocess(text):
    text = re.sub(r"<br\s*/?>", " ", text)      # do this first
    text = re.sub(r'[^a-zA-Z\s]', '', text)     # then this
    text = re.sub(r'\s+', ' ', text).strip()
    text = text.lower()
    tokens = word_tokenize(text)
    cleaned_tokens = [lemmatizer.lemmatize(t, pos='v') for t in tokens if t not in stop_words]
    return ' '.join(cleaned_tokens)
```

```python
df['clean_review'] = df['review'].apply(preprocess)
df['label'] = df['sentiment'].map({'positive': 1, 'negative': 0})

df.to_csv('cleaned_data.csv', index=False)
print(f"Saved {len(df)} cleaned rows to cleaned_data.csv")
```
